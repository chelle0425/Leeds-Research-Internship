```python
### GMT basemap ###

###########################################
## Specify coords
# llcrnrlat,llcrnrlon, urcrnrlat, urcrnrlon
# are the lat/lon values of the lower left and upper right corners of the map

llcrnrlon=-11 # lower left corner longitude 
llcrnrlat=49 # lower left corner latitude
urcrnrlon=3 # upper right corner longitude
urcrnrlat=61 # upper right corner latitude


proj_scale_w = 18 # specify projection scale (width)
proj_scale_unit = "c" # choose between c (cm) and i (inches)
# 17.78c = 7i

###########################################


def center(llcrnrlon, urcrnrlon, llcrnrlat, urcrnrlat):
    centerlon = (urcrnrlon - llcrnrlon)/2 + llcrnrlon # midpoints
    centerlat = (urcrnrlat - llcrnrlat)/2 + llcrnrlat
    return centerlon, centerlat

region=[llcrnrlon, urcrnrlon, llcrnrlat, urcrnrlat]



# grid = pygmt.datasets.load_earth_relief("30s", registration="gridline", region=region)


fig = pygmt.Figure()
pygmt.config(MAP_FRAME_TYPE="plain") # figure config (borders)
centerlon_hid, centerlat_hid = center(*region)


fig.coast(region=region,\
          projection="M%s/%s/%s%s" % (centerlon_hid, centerlat_hid, proj_scale_w, proj_scale_unit),\
          land="lightgrey", water="white",
          frame=["WSne", "a2f0.5"])

          # Mercator
          # a2f0.5 --> major ticks every 2 and minor ticks every 0.5

# fig.grdimage(grid=grid, cmap="bin/cpt/custom/seabluelandgrey.cpt")

fig.coast(shorelines="1/0.5p", borders="1/0.6p, mistyrose4,-")
# fig.coast(map_scale="g78.7/32+w200k", shorelines="1/0.5p", borders="1/0.6p, mistyrose4,-")


###########################################
### Plotting rivers ###

'''
pygmt.config(PROJ_ELLIPSOID="OSGB36")

x=[x] # must be list with one value or np array
y=[y]

if isinstance(region, str):
    region_str = region
elif isinstance(region, list):
    region_str = f"{region[0]}/{region[1]}/{region[2]}/{region[3]}"
else:
    raise Exception("invalid region input (must be either a str or a list)") 


with Session() as ses:
    with ses.virtualfile_from_vectors(x, y) as fin:
        args = [f'{fin}', f'-R{region_str}', f'-J{projection}', '-I', '-S']
        with GMTTempFile() as fout:
            ses.call_module(module="mapproject", args=' '.join(args)+ " ->" + fout.name)
            out = fout.read().strip()

lon, lat = [float(i) for i in out.split(' ')]

'''
fig.plot(data="data/WatercourseLink.gmt", pen="0.65p") #
fig.coast(lakes="turquoise3", rivers="a/1p,turquoise3,4_solid", water="turquoise3", transparency=30) # want water above national borders


###########################################

fig.show()
# fig.savefig("os_rivers.png")
```

```python
import geopandas as gpd


shapefile_path = "data/WatercourseLink.shp"

gdf_complete = gpd.read_file(shapefile_path)

gdf_complete['easting'] = gdf_complete.geometry.centroid.x
gdf_complete['northing'] = gdf_complete.geometry.centroid.y

csv_path = "data/WatercourseLink.csv"
gdf_complete.to_csv(csv_path, index=False)

```
